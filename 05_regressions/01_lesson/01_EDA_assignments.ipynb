```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

computers = pd.read_csv(r"C:\BIT_Data_analyst\05_regressions\01_lesson\Computers.csv")

computers.head()
```

```python
computers.describe().T
```

```python
plt.figure(figsize=(8,5))
sns.histplot(computers["price"], kde=True)
plt.title("Price histogram")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=computers["price"])
plt.title("Price boxplot")
plt.xlabel("Price")
plt.show()
```

```python
plt.figure(figsize=(10,4))
sns.histplot(computers["speed"], kde=True)
plt.title("Speed histogram")
plt.xlabel("Speed (MHz)")
plt.show()
```

```python

plt.figure(figsize=(8,5))
sns.histplot(computers["ram"], kde=True)
plt.title("RAM histogram")
plt.xlabel("RAM (MB)")
plt.show()
```

## Assignment 2: Variable Relationships

1. Build a correlation matrix of numeric features and pass it into a heatmap
2. Build a pairplot of numeric features in the data. Hint: df.select_dtypes("number") selects numeric columns. Consider sampling with df.sample(1000) to speed this up if it takes too long.
3. Build an lmplot of ram vs. price. Does being a premium model (premium == 'yes') impact this relationship? 

```python
plt.figure(figsize=(10,6))
sns.heatmap(computers.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation matrix")
plt.show()
```

```python
numeric = computers.select_dtypes("number")

sns.pairplot(numeric.sample(1000, random_state=1))
plt.show()
```

```python
sns.lmplot(
    data=computers,
    x="ram",
    y="price",
    hue="premium",
    height=8,
    aspect=1.2
)
plt.title("RAM vs Price with Premium Indicator")
plt.show()
```
