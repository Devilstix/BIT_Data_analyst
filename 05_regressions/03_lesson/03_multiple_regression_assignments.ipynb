## Assignment 2: Metrics

Calculate MSE and RMSE for your regression model fitted on all numeric features.

```python
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error
import numpy as np

# 1. Sukuriame X2 su visais skaitiniais požymiais
X2 = computers[['speed', 'hd', 'ram', 'screen', 'trend', 'ads']]
X2 = sm.add_constant(X2)

# 2. Tikslinė kintamoji
y = computers['price']

# 3. Sukuriame modelį
model2 = sm.OLS(y, X2).fit()

# 4. Prognozės
y_pred = model2.predict(X2)

# 5. MSE ir RMSE
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

mse, rmse

```
