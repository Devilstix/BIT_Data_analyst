8. Sukurkite list su pirkimo datomis tekstu (pvz. '2023-01-01') ir konvertuokite jas į datetime
objektus naudojant datetime.strptime.

```python
from datetime import datetime
sale_dates = ["2023-01-01", "2023-02-15", "2023-03-10"]
converted_dates = [datetime.strptime(data, "%Y-%m-%d") for data in sale_dates]
print(converted_dates)
```

9. Apskaičiuokite kiek dienų praėjo nuo kiekvienos pirkimo datos iki šiandienos
(datetime.date.today()) ir atspausdinkite.

```python
from datetime import datetime, date
sale_dates = ["2023-01-01", "2023-02-15", "2023-03-10"]
sales_dates_converted = [datetime.strptime(data, "%Y-%m-%d").date() for data in sale_dates]
today = date.today()
for data in sales_dates_converted:
    skirtumas = (today - data).days
    print(f"From {data} is {skirtumas} days apart")
```

10. Sukurkite žodyną, kuriame raktais yra produkto pavadinimai, o reikšmė – sąrašas mėnesinių
pardavimų. Raskite bendrą sumą kiekvienam produktui naudodami for ciklą ir sum().

```python
sales = { "Kompiuteris": [100, 120, 90, 150], "Telefonas": [200, 180, 220, 210], "Planšetė": [80, 95, 100, 85] }
for product, months in sales.items():
    total_sum = sum(months)
    print(f"{product}: {total_sum}")
```

# Funkcijų kūrimo užduotys (duomenų analitika):

1. Parašykite funkciją, kuri priima skaičių ir grąžina tą skaičių padaugintą iš 2

```python
def multiply(number):
    return number * 2

result = multiply(4)
print(result)
```

2. Sukurkite funkciją, kuri grąžina konversijos rodiklį (conversion rate), jei pateikiamas vartotojų
skaičius ir pirkėjų skaičius. Naudoti: (pirkėjai / vartotojai) * 100.

```python
def conversion_rate(users, buyers):
    return (buyers / users) * 100

rate = conversion_rate(200, 50)
print(rate)
```

3. Parašykite funkciją, kuri skaičiuoja CTR (click-through rate): paspaudimai / parodymai * 100

```python
def ctr(clicks, popups):
    return (clicks / popups) * 100

new_ctr = ctr(60, 80)
print(new_ctr)
```

4. Sukurkite funkciją, kuri apskaičiuoja vidutinę reikšmę iš pateikto sąrašo skaičių. Naudoti
sum() ir len().

```python
def average(numbers):
    return sum(numbers) / len(numbers)

data = [10, 20, 30, 40, 50]
result = average(data)
print(result)
```

5. Parašykite funkciją, kuri grąžina skirtumą tarp dviejų skaičių – naudinga apskaičiuoti mėnesio

```python
def difference(num1, num2):
    return num2 - num1

change = difference(3000, 4500)
print(change)
```

6. Sukurkite funkciją, kuri priima pelno ir išlaidų sąrašus ir grąžina bendrą pelną kiekvienam
laikotarpiui (pelno - išlaidos).

```python
def total_profit(profits, expenses):
    result = []
    for i in range(len(profits)):
        result.append(profits[i] - expenses[i])
    return result

profits = [1000, 1200, 900]
expenses = [400, 500, 300]
print(total_profit(profits, expenses))
```

8. Parašykite funkciją, kuri priima dict su pardavimais pagal regioną ir grąžina regioną su
didžiausia pardavimų suma. Naudoti max().

```python
def top_region(sales):
    return max(sales, key = sales.get)

sales_data = { "North": 1200, "South": 1500, "East": 900, "West": 1700 }
print(top_region(sales_data))
```

10. Sukurkite funkciją, kuri gauna list su datomis (tekstais) ir grąžina tų datų sąrašą kaip
datetime objektus. Naudoti datetime.strptime().

```python
from datetime import datetime
def convert_date(date_list):
    result = []
    for d in date_list:
        result.append(datetime.strptime(d, "%Y-%m-%d"))
    return result

dates = ["2024-05-10", "2025-01-29", "2026-12-31"]
converted = convert_date(dates)
print(converted)
```
